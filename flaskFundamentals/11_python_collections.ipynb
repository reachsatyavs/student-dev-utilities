## 1. `list` — an ordered, changeable collection

🎬 **Roll camera.**

Imagine you're the director. It's day one of casting for your next
blockbuster. You don't know who's walking through that door next — today
it's Yash, tomorrow maybe a fresh heroine joins, and by next week someone
might be shown the door entirely. The **order** people arrive in matters
too — whoever gets top billing goes first on the poster, always.

That, right there, is a **list**: a lineup that keeps whatever order you
built it in, and lets you add, remove, and rearrange people as the shoot
goes on. Cast changes, reshoots, last-minute additions — the list just
keeps rolling with the production.

**15 shots in this scene.** Guess the output before you run each cell, then
check the take.


### 🎬 Scene 1 — Building the cast

```python
# Take 1: the cast walks in, one by one, in the order they're signed
cast = ["Yash", "Sudeep"]
print(cast)
print(type(cast))
```

```python
# Take 2: "Who's my hero?" -- position 0 always answers that
print("Top-billed star:", cast[0])
print("Most recently signed:", cast[-1])
```

```python
# Take 3: a slice for the trailer -- just the first two names
print("Trailer credits:", cast[:2])
```

```python
# Take 4: "How big is this cast, exactly?" -- the producer always asks
print("Total cast size:", len(cast))
```

```python
# Take 5: CUT! A new actor signs mid-shoot -- straight in at the back of the line
cast.append("Puneeth Rajkumar")
print(cast)
```

### 🎬 Scene 2 — Rewrites and reshoots

```python
# Take 6: the producer insists the heroine gets TOP billing -- insert at position 0
cast.insert(0, "Rashmika Mandanna")
print(cast)
```

```python
# Take 7: the whole stunt team joins the credits at once -- .extend() adds a group
stunt_team = ["Stunt Silva", "Ravi Verma"]
cast.extend(stunt_team)
print(cast)
```

```python
# Take 8: CUT! Sudeep walks off the project -- .remove() drops the first match
cast.remove("Sudeep")
print(cast)
```

```python
# Take 9: the LAST name added gets cut for time -- .pop() removes and hands it back
dropped = cast.pop()
print("Dropped from credits:", dropped)
print("Remaining cast:", cast)
```

```python
# Take 10: "Wait, is Yash even in this movie?" -- and did we sign him twice?
print("Is Yash in the movie?", "Yash" in cast)
cast.append("Yash")  # oops, typo on the casting sheet
print("How many times does Yash appear now?", cast.count("Yash"))
```

### 🎬 Scene 3 — Wrap and roll credits

```python
# Take 11: the credits page must be A-to-Z, but billing order stays untouched
cast_sorted = sorted(cast)
print("Billing order:", cast)
print("Credits (A-Z):", cast_sorted)
```

```python
# Take 12: closing credits roll bottom-to-top -- reverse the list in place
cast.reverse()
print("Closing credits order:", cast)
cast.reverse()  # flip it back so the rest of the scene still makes sense
```

```python
# Take 13: the poster designer wants every name IN CAPS, "STARRING" and all
poster_credits = ["STARRING " + name.upper() for name in cast]
for line in poster_credits:
    print(line)
```

```python
# Take 14: a classic set mistake -- handing someone "the same" call sheet, not a copy
official_cast = cast
official_cast.append("Extra #47")
print("cast:", cast)  # changed too! it's the SAME list, not a copy

safe_copy = cast.copy()
safe_copy.append("Extra #48")
print("cast (unaffected this time):", cast)
print("safe_copy:", safe_copy)
```

```python
# Take 15 (that's a wrap): a full day's shot list -- a list OF lists
shot_list = [
    ["Wide shot", "Yash", "entrance"],
    ["Close-up", "Rashmika Mandanna", "reaction"],
    ["Action", "Puneeth Rajkumar", "car chase"],
]
for shot_type, actor, note in shot_list:
    print(f"{shot_type}: {actor} -- {note}")
```

## 2. `set` — an unordered collection of unique values

🎬 **New scene. New rules.**

You're now running the fan-club coordination desk — every big star has one.
Fans keep shouting genres at you: "Action!", "Action!", "Drama!", "Action!"
again. You don't care who said it twice, and you don't care what order they
shouted in — you just want to know: which genres, *distinct*, has this star
actually done? And can you answer "has he done a Comedy?" instantly, even
with a thousand fans shouting at once?

That's a **set**. No order, no duplicates, blazing-fast membership checks —
much like Rocky Bhai's own line, *"Whoever comes will go..."* — in a set,
whoever shows up twice only ever counts once.

**15 shots in this scene.**


### 🎬 Scene 1 — One of each, please

```python
# Take 1: two stars, and the genres each has worked in
yash_genres = {"Action", "Drama", "Period"}
puneeth_genres = {"Action", "Comedy", "Family", "Drama"}
print(yash_genres)
print(type(yash_genres))
```

```python
# Take 2: he signs on for a Thriller next
yash_genres.add("Thriller")
print(yash_genres)
```

```python
# Take 3: dropping a genre -- .remove() complains if it's missing, .discard() doesn't
yash_genres.remove("Period")
print(yash_genres)

yash_genres.discard("Period")  # already gone -- .discard() stays quiet about it
print(yash_genres)
```

```python
# Take 4: the fan club's favorite question
print("Has Yash done a Comedy?", "Comedy" in yash_genres)
print("Has Yash done a Thriller?", "Thriller" in yash_genres)
```

```python
# Take 5: how many DIFFERENT genres, total?
print("Distinct genres for Yash:", len(yash_genres))
```

### 🎬 Scene 2 — Comparing two fan clubs

```python
# Take 6: every genre either star has ever done
all_genres = yash_genres.union(puneeth_genres)
print("Every genre between the two:", all_genres)
```

```python
# Take 7: genres BOTH of them have done
common = yash_genres.intersection(puneeth_genres)
print("Genres both have done:", common)
```

```python
# Take 8: genres only Yash has done -- Puneeth hasn't touched these
only_yash = yash_genres.difference(puneeth_genres)
print("Genres only Yash has done:", only_yash)
```

```python
# Take 9: genres exactly ONE of them has done -- never both
not_shared = yash_genres.symmetric_difference(puneeth_genres)
print("Genres only one of them has done:", not_shared)
```

```python
# Take 10: "is every genre in this small set also one of Yash's?"
action_drama_only = {"Action", "Drama"}
print("Is {Action, Drama} a subset of Yash's genres?", action_drama_only.issubset(yash_genres))
print("Is Yash's genre set a superset of that?", yash_genres.issuperset(action_drama_only))
```

```python
# Take 11: two actors who have NEVER worked in the same genre -- true rivals
horror_only_actor = {"Horror"}
print("No overlap at all?", yash_genres.isdisjoint(horror_only_actor))
```

### 🎬 Scene 3 — Merging and cleaning up crowds

```python
# Take 12: raw fan votes for "Best Actor" -- full of repeats
votes = ["Yash", "Sudeep", "Yash", "Puneeth Rajkumar", "Sudeep", "Yash"]
unique_nominees = set(votes)
print("Raw votes:      ", votes)
print("Unique nominees:", unique_nominees)
```

```python
# Take 13: two fan clubs merge into one -- .update() changes the set in place
yash_fanclub_bangalore = {"Ravi", "Priya", "Manju"}
yash_fanclub_mysuru = {"Priya", "Kavya", "Deepak"}
yash_fanclub_bangalore.update(yash_fanclub_mysuru)
print("Combined fan club (no duplicate Priya!):", yash_fanclub_bangalore)
```

```python
# Take 14: every DISTINCT industry across a stack of movies -- in one line
movie_industries = ["Kannada", "Telugu", "Hindi", "Kannada", "Tamil", "Telugu"]
distinct_industries = {industry for industry in movie_industries}
print("Industries represented:", distinct_industries)
```

```python
# Take 15 (that's a wrap): a "locked" set -- once announced, nominees can't change
final_nominees = frozenset({"Yash", "Sudeep", "Puneeth Rajkumar"})
print(final_nominees)
try:
    final_nominees.add("Someone Else")
except AttributeError as error:
    print("Can't do that! Python says:", error)
```

## 3. `tuple` — an ordered, unchangeable collection

🎬 **Lock it in.**

Some decisions on a film set get made once, in a producers' meeting, and
then they are simply not up for discussion on set: the title, the release
year, the industry it releases in. Nobody — not the director, not the lead
actor, not even a very persuasive assistant director — gets to walk onto
set and quietly change the release year mid-shoot.

A **tuple** is that locked decision. It looks like a list, it's ordered
like a list, but once it's written inside `()`, it's *final*. That's not a
limitation — it's the entire point. It also unpacks beautifully into named
variables, so your code reads like a call sheet instead of a row of
mystery numbers.

**15 shots in this scene** — one ends in a very deliberate error, so read
it, don't fear it.


### 🎬 Scene 1 — The producer's decision

```python
# Take 1: locked in at the producers' meeting -- (title, year, industry)
kgf = ("KGF Chapter 2", 2022, "Kannada")
print(kgf)
print(type(kgf))
```

```python
# Take 2: reading the locked facts back
print("Title:", kgf[0])
print("Year:", kgf[1])
print("Industry:", kgf[2])
```

```python
# Take 3: the call sheet version -- names instead of positions
title, year, industry = kgf
print(f"{title} ({year}) - {industry} cinema")
```

```python
# Take 4: CUT! Someone tries to change the release year on set. Watch what happens.
try:
    kgf[1] = 2023
except TypeError as error:
    print("Can't do that! Python says:", error)
```

```python
# Take 5: how many facts did the producers actually lock in?
print("Facts locked in:", len(kgf))
```

### 🎬 Scene 2 — Scorecards and swaps

```python
# Take 6: a locked scorecard -- innings scores never get rewritten after the match
cricket_scores = (45, 82, 0, 45, 100, 45)
print("How many times did the score land on 45?", cricket_scores.count(45))
print("Position of the century (100):", cricket_scores.index(100))
```

```python
# Take 7: two locked facts, joined into one bigger locked fact
release_info = ("KGF Chapter 2", 2022)
box_office_info = ("Rs 1200 crore", "Blockbuster")
full_record = release_info + box_office_info
print(full_record)
```

```python
# Take 8: "Encore! Encore! Encore!" -- repeating a tuple with *
chant = ("Encore",) * 3
print(chant)
```

```python
# Take 9: the classic swap -- batting order changes without a spare variable
striker, non_striker = "Kohli", "Rohit"
print("Before:", striker, non_striker)

striker, non_striker = non_striker, striker  # tuple unpacking, under the hood
print("After: ", striker, non_striker)
```

### 🎬 Scene 3 — Gotchas and going nested

```python
# Take 10: a one-actor tuple needs a comma, or it's not a tuple at all
not_a_tuple = ("Yash")
actually_a_tuple = ("Yash",)
print(type(not_a_tuple))       # just a string!
print(type(actually_a_tuple))  # a real tuple
```

```python
# Take 11: a fixed seat allotment -- (row, seat) can't shift once printed on the ticket
vip_seat = ("Row A", 12)
premiere_ticket = ("KGF Chapter 2", vip_seat)
print(premiere_ticket)
print("Seat:", premiere_ticket[1])
```

```python
# Take 12: only a tuple can do this -- it's "lockable" enough to be a dict key
head_to_head = {
    ("India", "Australia"): "India won",
    ("India", "England"): "England won",
}
print(head_to_head[("India", "Australia")])
```

```python
# Take 13: a release calendar -- each entry locked, the calendar itself still growable
release_calendar = [
    ("KGF Chapter 2", 2022, "Kannada"),
    ("Pushpa 2", 2024, "Telugu"),
    ("Jawan", 2023, "Hindi"),
]
for title, year, industry in release_calendar:
    print(f"{year}: {title} ({industry})")
```

```python
# Take 14: sometimes you need it flexible for a moment, then lock it again
cast_tuple = ("Yash", "Sanjay Dutt", "Srinidhi Shetty")
cast_list = list(cast_tuple)         # unlock it
cast_list.append("Raveena Tandon")
cast_tuple_final = tuple(cast_list)  # lock it again -- final cast
print(cast_tuple_final)
```

```python
# Take 15 (that's a wrap): a function "returning two things" is secretly
# returning ONE tuple
def movie_stats(title, year):
    return title, year  # this is really: return (title, year)

result = movie_stats("KGF Chapter 2", 2022)
print(result, type(result))

name, release_year = result
print(f"{name} released in {release_year}")
```

## 4. `dict` — key/value pairs, looked up by name

🎬 **The full profile.**

A cast list tells you *who's in the movie*. It doesn't tell you Puneeth
Rajkumar's nickname, which industry he works in, or what he's most famous
for. Those facts don't have a natural "1st, 2nd, 3rd" order — there's no
sense in which "nickname" comes before "industry." What you actually want
is to ask for a fact **by name** and get it back instantly: "what's his
nickname?" → "Appu." No scanning, no counting positions.

That's a **dict**: every fact filed under a label you choose, so you pull
up exactly what you need the moment you ask for it — a well-organized
production office instead of a pile of loose papers.

**15 shots in this scene** — the last few build up to a full mini
production database.


### 🎬 Scene 1 — One actor's file

```python
# Take 1: an actor's production file
actor = {
    "name": "Puneeth Rajkumar",
    "nickname": "Appu",
    "industry": "Kannada",
    "famous_for": "Appu, Lucia, Yuvarathnaa",
}
print(actor)
print(type(actor))
```

```python
# Take 2: pulling one fact out, by name
print("Name:", actor["name"])
print("Nickname:", actor["nickname"])
```

```python
# Take 3: asking for a fact that might not be on file -- without crashing
print(actor.get("national_award"))
print(actor.get("national_award", "Not recorded"))
```

```python
# Take 4: reading the whole file, label by label
print("Keys:  ", actor.keys())
print("Values:", actor.values())

for key, value in actor.items():
    print(f"{key} -> {value}")
```

```python
# Take 5: how many facts are on file for this actor?
print("Facts on file:", len(actor))
```

### 🎬 Scene 2 — Updating the file

```python
# Take 6: new information comes in -- file it under a new label
actor["birth_year"] = 1975
print(actor)
```

```python
# Take 7: a whole batch of updates at once
actor.update({"home_state": "Karnataka", "debut_movie": "Appu"})
print(actor)
```

```python
# Take 8: retract one fact from the file, and keep a copy of what it was
removed_value = actor.pop("home_state")
print("Removed:", removed_value)
print(actor)
```

```python
# Take 9: "do we even have his birth year on file?"
print("Do we know his birth year?", "birth_year" in actor)
print("Do we know his blood group?", "blood_group" in actor)
```

```python
# Take 10: "if it's not on file yet, start it as empty -- otherwise leave it alone"
actor.setdefault("awards", [])  # only sets it if "awards" isn't already a key
actor["awards"].append("Karnataka State Film Award")
print(actor)
```

### 🎬 Scene 3 — Building the full production database

```python
# Take 11: a dict can hold a list, which can hold more dicts -- the exact shape
# MOVIES uses in 10_flask_movie_cast.py
movie = {
    "title": "KGF Chapter 2",
    "year": 2022,
    "cast": [
        {"role": "Hero", "name": "Yash"},
        {"role": "Antagonist", "name": "Sanjay Dutt"},
    ],
    "songs": ["Toofan", "Rocky Bhai"],
}
print(movie["title"], "cast:")
for member in movie["cast"]:
    print(" -", member["role"], ":", member["name"])
print("Songs:", ", ".join(movie["songs"]))
```

```python
# Take 12: build a lookup of "name -> length of name" in one line
cast_names = ["Yash", "Sanjay Dutt", "Srinidhi Shetty"]
name_lengths = {name: len(name) for name in cast_names}
print(name_lengths)
```

```python
# Take 13: merging two files into one, the modern way (Python 3.9+)
base_profile = {"name": "Yash", "industry": "Kannada"}
extra_facts = {"famous_for": "KGF, Rocking Star", "industry": "Kannada/Pan-India"}
full_profile = base_profile | extra_facts  # extra_facts wins on overlapping keys
print(full_profile)
```

```python
# Take 14: box office numbers, ranked -- sorting a dict by its VALUES
box_office_cr = {
    "KGF Chapter 2": 1200,
    "Kantara": 400,
    "Pushpa 2": 1800,
}
ranked = sorted(box_office_cr.items(), key=lambda item: item[1], reverse=True)
for title, collection in ranked:
    print(f"{title}: Rs {collection} crore")
```

```python
# Take 15 (that's a wrap): a whole mini database -- one dict, keyed by actor name
actor_database = {
    "Yash": {"industry": "Kannada", "famous_for": "KGF"},
    "Puneeth Rajkumar": {"industry": "Kannada", "famous_for": "Appu"},
    "Sudeep": {"industry": "Kannada", "famous_for": "Eega, Pailwaan"},
}
for name, profile in actor_database.items():
    print(f"{name} -> {profile['industry']} cinema, famous for {profile['famous_for']}")
```

## 🎬 That's a wrap — quick reference

| Type    | Ordered? | Changeable? | Duplicates allowed? | Syntax  | Best for |
|---------|----------|-------------|----------------------|---------|----------|
| `list`  | Yes      | Yes         | Yes                  | `[ ]`   | A growing, ordered lineup — a cast list |
| `set`   | No       | Yes (add/remove) | No              | `{ }`   | Unique values, instant "is it in here?" — genres, fan votes |
| `tuple` | Yes      | No          | Yes                  | `( )`   | Decisions locked in for good — `(title, year, industry)` |
| `dict`  | Yes (insertion order) | Yes | Keys: No, Values: Yes | `{key: value}` | A file you look up by name — an actor's profile |
